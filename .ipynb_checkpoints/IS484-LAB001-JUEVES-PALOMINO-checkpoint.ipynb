```python
def agente_termostato(temperatura_actual, temperatura_objetivo=22, margen=1.0):
    """
    Agente reactivo simple para controlar la temperatura de un ambiente.

    PEAS:
    Percepción: temperatura actual medida por un sensor.
    Acciones: activar aire acondicionado, activar calefacción
              o mantener el sistema sin cambios.
    Entorno: ambiente interior cuya temperatura puede variar.
    Objetivo: mantener la temperatura cercana a 22 °C.
    Medida de desempeño: mantener la temperatura dentro del margen
                         permitido y evitar cambios innecesarios.
    """

    diferencia = temperatura_actual - temperatura_objetivo

    if diferencia > margen:
        return "Activar aire acondicionado"

    elif diferencia < -margen:
        return "Activar calefacción"

    else:
        return "Mantener temperatura"
```

```python
percepciones = [
    18.0,
    19.5,
    21.8,
    24.3,
    26.0,
    23.0,
    20.5,
    21.0,  # Valor propio: límite inferior
    23.0   # Valor propio: límite superior
]

for t in percepciones:
    accion = agente_termostato(t)
    print(f"Percepción: {t} °C -> Acción del agente: {accion}")
```
